```python
timestamps = astro_3.Timestamps.to_numpy()[:3303].squeeze()
timestamps
```

```python
onsets, offsets = onep.find_onset_offset(astro_3.freeze_vector,astro_3.Timestamps)
```

```python
ax, across_eta_, time = onep.eta_individual_cells(
    data=sliced_dfz,
    timestamps=timestamps,
    events=[onsets],
    window=15
)
```

```python
max_indices = np.argmax(across_eta_, axis=1)
time = np.linspace(-5, 15, 5)
sorted_ind = np.argsort(max_indices)
sorted_arr = across_eta_[sorted_ind]
fig, ax = plt.subplots()
# ax.axvline(x=42, color='white', linestyle='--')
sns.heatmap(sorted_arr, cmap='mako',cbar=True, cbar_kws={"label": r"z-scored $\frac{dF}{F}$"})
ax.set_xticks(np.linspace(0, len(across_eta_.T), 5), time)
ax.set_title('Recall Cells: Astro3 Freeze Onset')
# ax.set_title('Recall Cells: Active FC-Recall Sorted from FC')
ax.set_xlabel('Time (seconds)')
ax.set_ylabel('Cell #')
```
