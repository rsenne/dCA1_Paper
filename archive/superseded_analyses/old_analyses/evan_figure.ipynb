```python
import bokeh.io
import matplotlib.pyplot

from onep import *
import holoviews as hv

hv.extension('bokeh')
from bokeh.plotting import show
#from onephoton import *
import holoviews as hv
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sb
import matplotlib as mpl
import statistics
import tifffile
import seaborn as sns
from scipy.stats import ttest_ind
```

```python
# initialize group lists
groups = [['astro3', 'astro4', 'astro5', 'astro6'], ['astro7', 'astro8', 'astro9']]

# initialize empty lists for overlap values ([0] is for Recall and [1] is for Generalization)
percent_overlap = [[], []]

# same as above but for SEM
SEM = []

# set indexer to -1 so the first loop will use 0 as the index and the second loop will use 1 as the index
index = -1

# loop for over both groups
for group in groups:
# increase the index value by 1 to align with respective group indexes
    index += 1
# loop over each animal in set group to calculate individual overlaps
for annie in group:
# create a variable for total cell population
    divisor = 0
# create a variable for overlapping cell population
    session_overlap = 0
# retrieve cell registration data for each animal for sessions n=2
    astro_tables = cell_registration.CellReg(annie, 'FOV1', 2)
# retrieve longitudinal cell registration csv for each animal for sessions n=2
    inds = astro_tables.load_registration_table().iloc[:, 0:2]
# retrieve longitudinal cell registration csv for each animal for sessions n=2
    inds = astro_tables.load_registration_table().iloc[:, 0:2]
# loop over each cell pairing to identify if there is a cell match across sessions
    for i in range(len(inds.iloc[:, 0])):
        if inds.iloc[:, 0][i] >= 0 and inds.iloc[:, 1][i] >= 0:
            session_overlap += 1
    if inds.iloc[:, 0][i] >= 0:
        divisor += 1
```

```python
# calculate the percent overlap for each animal across sessions
percent_overlap[index].append((session_overlap / divisor) * 100)
# df = pd.DataFrame(percent_overlap)
# df.to_csv(r"C:\Users\RamirezLab\Desktop\percent_overlaps.csv")
```

```python
# make the figure and axis
colors = ["#66C1A6", "#FC8C62"]
sns.set_style("ticks")
fig, ax = plt.subplots(figsize=(3, 4))
sns.set_palette(sns.color_palette(colors))

sns.barplot(percent_overlap, ci=68, capsize=.2, errwidth=1, alpha=0.8)


def width_changer(ax, new_val):
    for patch in ax.patches:
        cur_width = patch.get_width()


diff = cur_width - new_val
patch.set_width(new_val)
patch.set_x(patch.get_x() + diff * .5)

width_changer(ax, .4)

sns.swarmplot(percent_overlap, size=4, linewidth=0.3, edgecolor='black')

# remove extra legend handles
ax.set_xticklabels(['FC vs Recall', 'FC vs Generalization'])
ax.set_ylabel("Overlap %")
ax.set_ylim(0, 70)
sns.despine()
plt.show()

#Statistics
# calculate SEM values for both groups
for x in percent_overlap:
    SEM.append(statistics.stdev(x) / (len(x) ** 0.5))

# independent samples t-test
t_statistic, p_value = ttest_ind(percent_overlap[0], percent_overlap[1])
t_test_result = f"t-test: t = {t_statistic:.2f}, p = {p_value:.4f}"
print(t_test_result)

# # display t-test result on graph
# ax.annotate(t_test_result, xy=(0.5, 0.95), xycoords='axes fraction', ha='center', fontsize=10, color='black')
```
