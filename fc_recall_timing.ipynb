```python
#astro_3_fc = onep.InscopixProcessing(session='fc', animal='astro10',data_directory='/Users/suthardr/Desktop/')

astro_3_recall = onep.InscopixProcessing(session='fc', animal='astro3', data_directory='/Users/suthardr/Desktop/')

collection = onep.dCA1Group(astro_3_recall)
collection.preprocess()
```

```python
#Get registration table
table = astro_3_recall.load_registration_table(session_subset=[0,1],filter_accepted=True)
```

```python
#Filter by cells only in both days
both_idxs = table.loc[(table[0]!=-1)&(table[1]!=-1)]
```

```python
both_idxs
```

```python
fc_traces = astro_3_fc.get_traces()
recall_traces = astro_3_recall.get_traces() #important to use this function and not index into object.accepted traces bc 'table' is in reference to all the traces.
```

```python
fc_active = fc_traces[both_idxs[0]]
fc_active
```

```python
recall_active = recall_traces[both_idxs[1]]
recall_active
```

```python
fc_dfz = stats.zscore(fc_active, axis=1)
fc_dfz
```

```python
recall_dfz = stats.zscore(recall_active, axis=1)
recall_dfz
```

```python
fc_dfz = fc_dfz[:,:3303]
fc_dfz
```

```python
recall_dfz = recall_dfz[:,:3303]
recall_dfz
```

```python
ax, across_eta_ = onep.eta_individual_cells(data=recall_dfz, timestamps=astro_3_fc.timestamps[:3303], events=[[120,180,240,300], ], window=15)
```

```python
max_idxs = np.zeros(23)
for i in range(across_eta_.shape[0]):
    idx = np.argmax(across_eta_[i])
    max_idxs[i] = idx
```

```python
new_table = np.hstack((both_idxs, max_idxs.reshape(-1, 1)))
sorted_idxs = np.argsort(new_table[:, 2])
```

```python
max_indices = np.argmax(across_eta_, axis=1)
time = np.linspace(-5, 15, 5)
# sorted_ind = np.argsort(max_indices)
sorted_arr = across_eta_[sorted_idxs]
fig, ax = plt.subplots()
ax.axvline(x=42, color='white', linestyle='--')
sns.heatmap(sorted_arr, cmap='mako',cbar=True, cbar_kws={"label": r"z-scored $\frac{dF}{F}$"})
ax.set_xticks(np.linspace(0, len(across_eta_.T), 5), time)
#ax.set_title('FC Cells: Active FC-Recall Sorted from FC')
ax.set_title('Recall Cells: Active FC-Recall Sorted from FC')
ax.set_xlabel('Time (seconds)')
ax.set_ylabel('Cell #')
fig.savefig('/Users/suthardr/Desktop/Astro9_recall_fc_sorted.png')
#fig.savefig('/Users/suthardr/Desktop/Astro9_fc_fc_sorted.png')
```
