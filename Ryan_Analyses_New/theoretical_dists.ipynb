```python
import numpy as np
import matplotlib.pyplot as plt

T = 2.0
dt = 0.01
time = np.arange(0, T, dt)
t0 = 1.0                 # true underlying peak time
n_neurons = 500
n_trials = 20

# Single-peaked event-locked rate function, same for all neurons
sigma = 0.1
lam = 5.0 * np.exp(-0.5 * ((time - t0) / sigma)**2)  # spikes/s

peak_times_null = []

for i in range(n_neurons):
    psth = np.zeros_like(time)
    for tr in range(n_trials):
        # Inhomogeneous Poisson spikes for this neuron & trial
        spikes = np.random.poisson(lam * dt)
        psth += spikes
    psth = psth / n_trials
    # Estimate peak time as argmax of PSTH
    peak_idx = np.argmax(psth)
    peak_times_null.append(time[peak_idx])

peak_times_null = np.array(peak_times_null)
```

```python
plt.hist(peak_times_null, alpha=0.5, label='Gaussian Latencies')
```

```python
T = 2.0
dt = 0.01
time = np.arange(0, T, dt)
n_neurons = 500
n_trials_hom = 3  # keep small so PSTH retains randomness

# Homogeneous (time-independent) rate function, same for all neurons
lam_hom = 5.0 * np.ones_like(time)  # spikes/s

peak_times_hom = []

for i in range(n_neurons):
    psth = np.zeros_like(time)
    for tr in range(n_trials_hom):
        # Homogeneous Poisson spikes for this neuron & trial
        spikes = np.random.poisson(lam_hom * dt)
        psth += spikes
    psth = psth / n_trials_hom

    # Estimate peak time as argmax of PSTH, with random tie-breaking
    maxval = psth.max()
    candidate_idxs = np.where(psth == maxval)[0]
    peak_idx = np.random.choice(candidate_idxs)
    peak_times_hom.append(time[peak_idx])

peak_times_hom = np.array(peak_times_hom)

# Optional: compare histograms of the two nulls
plt.figure(figsize=(8, 4))
plt.hist(peak_times_null, bins=20, alpha=0.5, density=True, label="Event-locked")
plt.hist(peak_times_hom, bins=20, alpha=0.5, density=True, label="Homogeneous Point Process")
plt.xlabel("Peak time (s)")
plt.ylabel("Density")
plt.legend()
plt.tight_layout()
plt.show()
```
