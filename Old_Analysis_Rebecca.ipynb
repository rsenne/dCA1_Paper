```python
timestamps = astro_4.Timestamps.to_numpy()[:3303].squeeze()
timestamps
```

```python
# Verify the range of timestamps
max_time = max(timestamps)
min_time = min(timestamps)

print(f"Max timestamp: {max_time}")
print(f"Min timestamp: {min_time}")
```

```python
# Define your window
window =2

# Clip offsets to ensure offset + window is within the valid range
clipped_offsets = []
for offset in offsets:
    adjusted_offset = min(max(offset, min_time - window), max_time - window)
    clipped_offsets.append(adjusted_offset)

# Print the clipped offsets
print(f"Clipped offsets: {clipped_offsets}")

# Verify x_new values
x_new = [offset + window for offset in clipped_offsets]
print(f"x_new: {x_new}")

# Ensure x_new values are within the timestamps range
if any(val > max_time or val < min_time for val in x_new):
    raise ValueError(f"x_new values out of range: {x_new}")
```
