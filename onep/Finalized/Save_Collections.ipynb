```python
import os
import onep

base_dir = r"/Volumes/rsenne/dCA1_Clean_Data"

mouse_ids = ['astroF3', 'astroF5', 'astroF6', 'astroF7', 'astroF8', 'astroF9', 'astroF10','astroM3', 'astroM4', 'astroM5', 'astroM6', 'astroM7', 'astroM8', 'astroM9', 'astroM10']
# sessions = ['hab', 'fc', 'cxta', 'cxtb']
sessions = ['cxta', 'cxtb']
collection = []

for mouse_id in mouse_ids:
    for session in sessions:
        # Build the expected file path
        file_path = os.path.join(
            base_dir, 'Cell_Traces', f'{mouse_id}_traces', f'{mouse_id}_{session}_traces.csv'
        )
        if os.path.exists(file_path): #if this filepath exists for that specific mouse, run preprocessing
            try:
                obj = onep.InscopixProcessing(animal=mouse_id, session=session, data_directory=base_dir)
                collection.append(obj)
                print(f"Loaded: {mouse_id} - {session}")
            except Exception as e:
                print(f"Error loading {mouse_id} {session}: {e}")
        else:
            print(f"Missing file: {file_path}")
```

```python
for obj in collection:
    print(obj.animal, obj.session)
```

```python
import pickle
from collections import defaultdict

session_groups = defaultdict(list)

for obj in collection:
    session_groups[obj.session].append(obj)

for session, mice in session_groups.items():
    group = onep.dCA1Group(*mice)
    group.preprocess()
    path = f"/Volumes/rsenne/dCA1_Clean_Data/Dill/collection_{session}_allmice.pkl"
    # path = f"/Users/suthardr/Desktop/collection_{session}_allmice.pkl"
    with open(path, 'wb') as f:
        pickle.dump(group, f)
    print(f"Pickled group for {session}: {path}")
```
