```python
def resize_images(images):
    rows = [im.shape[0] for im in images]
    cols = [im.shape[1] for im in images]
    i = np.min(rows)
    j = np.min(cols)
    resized = [im[0:i,0:j] for im in images]
    return resized
```

```python
def plot_overlaid_rgb(im1,im2,scale_factor=3,gain=5):
    opts_im = opts.Image(width=int(im1.shape[1])*scale_factor,height=int(im1.shape[0])*scale_factor,alpha=.8)
    # opts = opts.Image(width=int(np.min([im1.shape[1],im2.shape[1]]))*scale_factor,height=int(np.min([im1.shape[0],im2.shape[0]]))*scale_factor,
    #                       cmap=cmap2,alpha=.5)
    rgb = hv.RGB(np.dstack([im1*gain,im2*gain,np.zeros((im1.shape[0],im1.shape[1]))])).opts(width=298*3,height=197*3)
    return rgb
```

```python
plot_overlaid_rgb(im1,im2)
```

```python
def plot_translate(im1,im2,shift_x,shift_y,rotation,shear):
    tform = AffineTransform(scale=(1.0, 1.0), rotation=rotation, shear=shear,
                        translation=(shift_x,shift_y))
    im2_t = warp(im2, tform.inverse)
    return plot_overlaid_rgb(im1,im2_t)
```
