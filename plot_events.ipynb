```python
import matplotlib.pyplot as plt
import pickle 
import numpy as np
import scipy.stats as stats
import holoviews as hv
hv.extension('bokeh')
```

```python
file = '/Users/amonast/Desktop/dCA1_astro/Analysis/Deconvolution/astro4_fc_traces_l0deconv.pkl'
data = np.load(file,allow_pickle=True)
dff = data['dff']
model = data['est']
T = dff.shape[1]
```

```python
def plot_deconv(data, i, T, frame_rate=10):
    '''
    works best in jupyter notebook:
    import holoviews as hv
    hv.extension('bokeh')

    S: dict with keys output from deconvolution: {'dff': array, dff traces NxT neurons by time
                        'estimated_calcium': array, NxT smoothed model estimate of calcium from deconvolution
                        'event_mag': list, len N Neurons, each item is that cell's event magnitudes
                        'spiketimes': list, len N Neurons, each item is that cell's event times (frames) }
    i: indices of cells (optional), array or single int
    T: total number of frames for all traces (should match up with time dimension of S.dff and S.estimated_calcium)
    frame_rate: frame rate for imaging, default 30 fps

    returns:
        plot: holoviews object with traces and events all plotted together.
    '''
    est = data['est'][i]
    dff = data['dff'][i]
    event_mag_pos = np.squeeze(data['events'][i])
    spikes = np.squeeze(data['times'][i])

    event_mag = np.zeros(T)
    event_mag[spikes] = event_mag_pos

    event_plot = spikes, -.01 * np.ones(spikes.shape)
    t = T / frame_rate

    plot = hv.Curve(dff, label='Raw df/f').opts(width=900, height=300, ylabel='df/f', xlabel='Frame #') * \
           hv.Curve(est, label='Deconvolution model fit').opts(width=900, height=300, color='#e5ae38') * \
           hv.Curve(event_mag, label='Deconvolved activity').opts(width=900, height=300, color='red') * \
           hv.Scatter(event_plot, label='Spikes').opts(fill_color='k', line_color='k', size=3)
    plot.opts(legend_position='right')
    return plot
```
