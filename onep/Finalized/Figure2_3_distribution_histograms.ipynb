```python
mpl.rcParams['svg.fonttype'] = 'none'
mpl.rcParams['font.sans-serif'] = ['Arial']
mpl.rcParams['font.family'] = 'sans-serif'
mpl.rcParams['text.usetex'] = False
```

```python
plt.figure()
fig, ax = plt.subplots(figsize=(3.5, 3.5), dpi=600)
sns.set(style="ticks", font="Arial")
ax.set_xlim([-1,50])
ax.set_xlabel('Time (seconds)', fontsize=10, fontname='Arial')
ax.set_ylabel('Density',fontsize=10)
ax.tick_params(labelsize=8)
ax.grid(False)
sns.despine(ax=ax)

#sns.distplot(data_A['pred_mu'], color="#75BFAA", label="CxtA", hist_kws={'alpha':.7}, kde_kws={'linewidth':3})
#sns.distplot(data_B['pred_mu'], color="black", label="CxtB", hist_kws={'alpha':.7}, kde_kws={'linewidth':3})
sns.distplot(data_FC['pred_mu'], color="grey", label="FC", hist_kws={'alpha':.7}, kde_kws={'linewidth':1})
plt.ylim(0, 0.30)
plt.tight_layout(pad=3)
ax.set_title('Distribution of Estimated Peaks',fontsize=10, pad=10, fontname='Arial')
ax.set_xlabel('Peak (mu; seconds)',fontsize=10)
ax.legend()
fig.savefig("/Users/suthardr/Desktop/Revision/FC_Groups_histogram_predictedpeaks.svg", format="svg", bbox_inches="tight")
```
