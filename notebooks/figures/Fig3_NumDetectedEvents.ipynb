```python
import seaborn as sns
import matplotlib.pyplot as plt

def plot_num_events_box(df, title="", save_path=None):
    sns.set(style="ticks", font="Arial")
    plt.rcParams.update({
        'font.size': 8,
        'svg.fonttype': 'none'   # <-- keep text editable in SVG
    })

    fig, ax = plt.subplots(figsize=(1.5, 2), dpi=600)

    palette = {
        'CxtA': '#66C1A6',  # teal
        'CxtB': 'black'     # black
    }

    sns.boxplot(
        x='Context', y='Num_Events', data=df,
        width=0.5,
        showcaps=True,
        boxprops=dict(alpha=0.7),
        medianprops={'color': 'white', 'linewidth': 1},
        whiskerprops={'linewidth': 1, 'color': 'black'},
        capprops={'linewidth': 1, 'color': 'black'},
        fliersize=0,
        palette=palette,
        ax=ax
    )

    sns.stripplot(
        x='Context', y='Num_Events', data=df,
        jitter=True,
        linewidth=0.3,
        edgecolor='white',
        palette=palette,
        size=3,
        ax=ax
    )

    ax.legend_.remove() if ax.legend_ else None
    ax.set_ylabel('# Detected Events', fontsize=10)
    ax.set_xlabel('', fontsize=10)
    ax.tick_params(labelsize=8)
    ax.set_ylim(0, 10)
    ax.set_title(title, fontsize=10, pad=10)
    sns.despine()
    plt.tight_layout()

    if save_path:
        plt.savefig(save_path, dpi=600, bbox_inches='tight', format='svg')

    plt.show()
    return fig, ax

plot_num_events_box(df, save_path='/Users/suthardr/Desktop/Revision/fig3/num_events_Fig3.svg')
```

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

df["Sex"] = df["Mouse"].astype(str).str.extract(r'([FM])', expand=False)

sns.set(style="ticks", font="Arial")
plt.rcParams.update({
    'font.size': 8,
    'svg.fonttype': 'none'
})

context_order = sorted(df["Context"].unique())  #
sex_order = ["F", "M"]

sex_palette = {
    "F": "#e377c2",
    "M": "#1f77b4"
}

fig, ax = plt.subplots(figsize=(2, 2.5), dpi=600)

#   BOX PLOT (GROUPED)
sns.boxplot(
    data=df,
    x="Context",
    y="Num_Events",
    hue="Sex",
    order=context_order,
    hue_order=sex_order,
    palette=sex_palette,
    width=0.5,
    fliersize=0,
    linewidth=0.8,
    boxprops=dict(alpha=0.7),
    ax=ax
)

#  STRIP PLOT ON TOP
sns.stripplot(
    data=df,
    x="Context",
    y="Num_Events",
    hue="Sex",
    order=context_order,
    hue_order=sex_order,
    palette=sex_palette,
    jitter=True,
    size=3,
    edgecolor="white",
    linewidth=0.3,
    ax=ax
)

# Remove duplicate legends from box + strip
handles, labels = ax.get_legend_handles_labels()
ax.legend(
    handles[:2], labels[:2],
    fontsize=8,
    title="Sex",
    title_fontsize=8,
    loc="upper right",
    frameon=False
)

#  AXIS FORMATTING
ax.set_ylabel("# Detected Events", fontsize=10)
ax.set_xlabel("", fontsize=10)
ax.tick_params(labelsize=8)
ax.set_ylim(0, 8)
sns.despine()
plt.tight_layout()

plt.savefig(
    '/Users/suthardr/Desktop/Revision/fig3/num_events_Fig3_sex_context_boxplot.svg',
    dpi=600,
    bbox_inches='tight',
    format='svg'
)

plt.show()
```

```python
import statsmodels.formula.api as smf
import statsmodels.api as sm

# make sure Context + Sex are categorical
df["Context"] = df["Context"].astype("category")
df["Sex"] = df["Sex"].astype("category")

model = smf.ols("Num_Events ~ Context * Sex", data=df).fit()
anova_table = sm.stats.anova_lm(model, typ=2)

print(model.summary())
print("\nANOVA Table:")
print(anova_table)
```
